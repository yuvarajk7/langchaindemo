### Difference between % and !

| **Feature**              | **% (Magic Commands)**                                                                 | **! (Shell Commands)**                                                                 |
|--------------------------|----------------------------------------------------------------------------------------|----------------------------------------------------------------------------------------|
| **Execution Context**    | Executes commands within the IPython kernel's process.                                | Executes commands by spawning a new, temporary subprocess (a shell).                   |
| **State Persistence**    | Can affect the state of your notebook's kernel. <br> E.g., `%cd` changes directory persistently. | Changes are not persistent. <br> E.g., `!cd` affects only the temporary subprocess.   |
| **Standard Commands**    | IPython has "magic" versions of common shell commands (e.g., `%ls`, `%cd`). Preferred for interacting with notebooks. | Executes any terminal command (e.g., `!ls`, `!echo`, `!git clone`).                   |
| **Package Management**   | `%pip install` ensures installation into the correct virtual environment used by the kernel. | `!pip install` may install packages incorrectly or outside the current environment.   |
| **Command Types**        | Includes IPython-specific features like `%timeit`, `%matplotlib inline`, `%who`, etc. | Limited to shell-executable commands.                                                  |
| **Output**               | Can provide structured, colorized, and context-aware output.                         | Returns output as plain text.                                                         |
